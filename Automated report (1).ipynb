```python
import pandas as pd
data = pd.read_csv("company1.csv")
print(data)
```

```python
print(data.isnull().sum())
```

```python
import numpy as np
mean = data["salary"].mean()
print(mean)
```

```python
data["salary"] = data["salary"].replace(np.nan, mean)
```

```python
print(data)
```

```python
print(data.fillna(method = "bfill")) #backward fill 
```

```python
print(data.fillna(method = "ffill")) #forward fill
```

```python
report = data.describe()

report.to_excel("summary_report.xlsx")

print("Report Generated Successfully")
```

```python
#sales trend graph
import matplotlib.pyplot as plt

plt.plot(data['salary'])

plt.title("Salary Trend")

plt.show()
```
